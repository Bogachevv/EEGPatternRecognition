```python
import numpy as np
from omegaconf import OmegaConf

from tqdm.notebook import tqdm
```

```python
np.random.seed(42)

adj_seed_ls = np.random.choice(
    a=np.logspace(base=2, start=8, stop=16, num=1_000, dtype=int),
    size=16,
    replace=False
)

weights_seed_ls = np.random.choice(
    a=np.logspace(base=2, start=8, stop=16, num=1_000, dtype=int),
    size=16,
    replace=False
)

print(f"{adj_seed_ls=}\n\n{weights_seed_ls=}")
```

```python
cfg_path=r'C:\Users\Vladimir\PycharmProjects\EEGPatternRecognition\configs\bs_cfg.yaml'
work_cfg_path=r'C:\Users\Vladimir\PycharmProjects\EEGPatternRecognition\configs\work_cfg.yaml'

for weights_seed in tqdm(weights_seed_ls):
    for adj_seed in tqdm(adj_seed_ls):
        cfg = OmegaConf.load(cfg_path)
        cfg['seed'] = weights_seed
        cfg['adj_seed'] = adj_seed
        cfg['run_name'] = r'random_init'

        with open(work_cfg_path, 'w') as f:
            print(
                OmegaConf.to_yaml(cfg, resolve=True),
                file=f
            )

        run(
            cfg_path=work_cfg_path
        )
```
